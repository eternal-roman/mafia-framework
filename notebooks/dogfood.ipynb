# Mafia framework dogfood

Offline contract check. No network. No personal book. Confirms the public skill is short, complete, and runnable.


```python
from pathlib import Path
import json, subprocess, sys
ROOT = Path.cwd().resolve()
assert (ROOT / "SKILL.md").exists(), "run from the repo root"
print("root", ROOT.name)
skill = (ROOT / "SKILL.md").read_text()
lines = skill.count(chr(10)) + 1
words = len(skill.split())
print("SKILL.md lines", lines)
print("SKILL.md words", words)
assert lines < 120, "SKILL.md should stay short"
```

```python
r = subprocess.run([sys.executable, str(ROOT / "scripts/selftest.py")], capture_output=True, text=True)
print(r.stdout)
assert r.returncode == 0, r.stdout + r.stderr
p = subprocess.run([sys.executable, str(ROOT / "scripts/pii_scan.py")], capture_output=True, text=True)
print(p.stdout)
assert p.returncode == 0, p.stdout + p.stderr
```

```python
fixture = (ROOT / "examples/format.md").read_text()
required = ["Key Roles", "How the Game Is Going", "Townsfolk Interests", "Who Is Winning", "Next-Best Decisions", "Evidence Ledger"]
missing = [s for s in required if s not in fixture]
assert not missing, missing
evals = json.loads((ROOT / "evals/evals.json").read_text())
triggers = json.loads((ROOT / "evals/triggers.json").read_text())
assert len(evals["test_cases"]) >= 3
assert len(triggers["test_cases"]) >= 3
assert any(not c["should_trigger"] for c in triggers["test_cases"])
assert any(c["should_trigger"] for c in triggers["test_cases"])
print("fixture sections", len(required))
print("evals", len(evals["test_cases"]), "triggers", len(triggers["test_cases"]))
print("DOGFOOD PASS")
```
